```python
from agents.agents import route_agent, weather_agent, online_search_agent, code_agent, main_agent
from agents.home_assistant import HomeAssistant
from langchain_core.messages import (
    HumanMessage,
)
```

```python
assistant = HomeAssistant(memory_timeframe=5, quirky_response_time=3)  # memory timeframe in minutes
```

```python
async def main():
    async for msg in  assistant.get_answer("What's the weather like today in New York City?", tone='sassy', language='English'):
        print(msg)


await main()
```

```python
for question in [
    "Who won the nobel prizes in 2025 in physics ?",
    "I want to go to dorval airport, what highway should I take?",
    "Should I wear a raincoat tomorrow ?",
    "Who won the oscar for the best actress in 2024 ?",
    "Calculate 123563523 * 13451134",
    "What highway do I take to go to dorval airport and should I wear a raincoat tomorrow ?"
]:
    display(assistant.get_answer(question))
```
